```python

    
FIELDS = ["displayName", "chapter"]
REQUIRED_KEY = "displayName"  # only keep objects that contain this key

# A JS double-quoted string literal, with escape support
JS_STR = r'"(?:[^"\\]|\\.)*"'

# Start of any $R[n]={ object
ANCHOR_RE = re.compile(r'\$R\[(\d+)\]=\{')

# Flat object body: anything except braces/quotes, or whole string literals
BODY_RE = re.compile(r'(?:[^{}"]|' + JS_STR + r')*')

# key:value pairs. The lookbehind ensures "displayName" doesn't match inside
# "authorDisplayName" and "chapter" doesn't match inside "authorChapter".
KV_RE = re.compile(
    r'(?<![\w$])(\w+):(' + JS_STR + r'|!0|!1|null|void 0|-?\d+(?:\.\d+)?)'
)

    
def js_unescape(s):
    """Decode a quoted JS string literal into a Python str."""
    s = re.sub(r'\\x([0-9a-fA-F]{2})', r'\\u00\1', s)  # \x3C -> \u003C
    s = s.replace("\\'", "'")
    try:
        # strict=False allows literal newlines/tabs/control chars inside strings
        return json.loads(s, strict=False)
    except json.JSONDecodeError:
        # Fallback: strip the quotes and return the raw contents
        return s[1:-1]


def parse_value(v):
    if v.startswith('"'):
        return js_unescape(v)
    if v == "!0":
        return True
    if v == "!1":
        return False
    if v in ("null", "void 0"):
        return None
    return float(v) if "." in v else int(v)


def extract_proposal_page(js_text, fields=FIELDS, required_key=REQUIRED_KEY):
    results = []
    for m in ANCHOR_RE.finditer(js_text):
        body = BODY_RE.match(js_text, m.end()).group(0)
        parsed = {k: parse_value(v) for k, v in KV_RE.findall(body)}
        if required_key not in parsed:
            continue
        row = {"ref": int(m.group(1))}
        row.update({f: parsed.get(f) for f in fields})  # None if missing/truncated
        results.append(row)
    return pd.DataFrame(results)


cosponsors = extract_proposal_page(js_text)
```

## proposal page parsing

```python

all_cosponsors = pd.DataFrame()

for index in range(proposals_df.shape[0]):

    proposal_id = proposals_df.iloc[index,]['proposalId']
    proposal_url = 'https://convention.socialists.nyc/proposals/' + proposal_id
    
    proposal_title = proposals_df.iloc[index,]['title']
    
    response = requests.get(proposal_url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")

    script_element = soup.find("script", id="$tsr-stream-barrier")

    js_text = script_element.text
    cosponsors = extract_proposal_page(js_text)
    cosponsors['proposal_title'] = proposal_title
    cosponsors['proposal_id'] = proposal_id

    print(cosponsors)
    all_cosponsors = pd.concat([all_cosponsors, cosponsors], axis=0)
```

```python

jaccards = pd.DataFrame()
unique_proposals = np.unique(all_cosponsors['proposal_title'])
for prop_x in unique_proposals:
    x_set = set(all_cosponsors[all_cosponsors['proposal_title'] == prop_x]['displayName'])
    for prop_y in unique_proposals:
        y_set = set(all_cosponsors[all_cosponsors['proposal_title'] == prop_y]['displayName'])
        
        intersection = len(x_set.intersection(y_set))
        union = len(x_set.union(y_set))
        jaccard_sim = intersection / union if union > 0 else 0.0

        chunk = pd.DataFrame(data={'prop_x':[prop_x], 'prop_y':[prop_y], 'jaccard':[jaccard_sim]})
        print(chunk)
        jaccards = pd.concat([jaccards, chunk], axis=0)
jaccards
        
    
```

```python
G_jaccard = nx.from_pandas_edgelist(
    jaccards[jaccards['jaccard'] > .25], 
    source='prop_x', 
    target='prop_y',
    edge_attr=['jaccard']
)
```

```python
nx.write_gml(G_jaccard, 'dsa/network_jacc.gml')
```

#### Alternative topologies based on jaccard similarity thresholds 

```python
jaccards = pd.DataFrame()
unique_proposals = np.unique(all_cosponsors['proposal_title'])
for prop_x in unique_proposals:
    x_set = set(all_cosponsors[all_cosponsors['proposal_title'] == prop_x]['displayName'])
    for prop_y in unique_proposals:
        y_set = set(all_cosponsors[all_cosponsors['proposal_title'] == prop_y]['displayName'])
        
        intersection = len(x_set.intersection(y_set))
        union = len(x_set.union(y_set))
        jaccard_sim = intersection / union if union > 0 else 0.0

        chunk = pd.DataFrame(data={'prop_x':[prop_x], 'prop_y':[prop_y], 'jaccard':[jaccard_sim]})
        jaccards = pd.concat([jaccards, chunk], axis=0)
jaccards
        
    
```

## by people now 

```python
jaccards_person = pd.DataFrame()
unique_people = np.unique(all_cosponsors['displayName'])
for peep_x in unique_people:
    asdf = pd.DataFrame(all_cosponsors[all_cosponsors['displayName'] == peep_x]['proposal_title'])
    asdf['peep_x'] = peep_x

    df_merge = pd.merge(asdf, all_cosponsors, on='proposal_title')
    intersection_df = df_merge.groupby(['peep_x','displayName']).count().reset_index()
    intersection_df = intersection_df[['peep_x','displayName','proposal_title']].rename({'proposal_title':'intersection'}, axis=1)
    intersection_df['all_a'] = all_cosponsors[all_cosponsors['displayName'] == peep_x]['proposal_title'].nunique()
    
    all_counts = all_cosponsors.groupby('displayName')['proposal_title'].nunique().rename('all_b')
    intersection_df = pd.merge(intersection_df, all_counts, on='displayName')
    intersection_df['jacc'] = intersection_df['intersection'] / (intersection_df['all_a'] + intersection_df['all_b'] - intersection_df['intersection'])
    print(intersection_df)
    jaccards_person = pd.concat([jaccards_person, intersection_df], axis=0)
```
